```python
#1 WRITE A PROGRAM TO ADD A KEY AND VALUE TO THE DICTIONARY
my_dict={0:10,1:20}
my_dict[2]=30
print("Dictionary:",my_dict)
```

```python
#2 WRITE A PROGRAM TO CONCATENATE THE FOLLOWING DICTIONARIES TO CREATE A NEW ONE
dic1={1:10,2:20}
dic2={3:30,4:40}
dic3={5:50,6:60}
new_dict={}
new_dict.update(dic1)
new_dict.update(dic2)
new_dict.update(dic3)
print(new_dict)
```

```python
#3 WRITE A PROGRAM TO CHECK IF THE GIVEN KEY EXIST IN A DICTIONARY
my_dict={101:"Drishti",102:"Sakshi",103:"Ishmeet"}
key=int(input("Enter key to search:"))
if key in my_dict:
    print("Key exist in the dictionary")
else:
    print("Key does not exist in the dictionary")
```

```python
#4 WRITE A PROGRAM TO ITERATE OVER A DICTIONARY USING FOR LOOP AND PRINT THE KEYS ALONE,ALONE AND BOTH KEY VALUES
my_dict={
    101:"Drishti",
    102:"Sakshi",
    103:"Ishmeet"
}
print("Keys:")
for key in my_dict:
    print (key)
for values in my_dict.values():
    print(values)
    print("\nKeys and values:")
for key,value in my_dict.items():
    print(key,value)
```

```python
#5 WRITE A PROGRAM TO PREPARE A DICTIONARY WHERE THE KEYS ARE NUMBERS BETWEEN 1 AND 15(BOTH INCLUDED) AND THE VALUES ARE SQUARES OF THE KEYS
my_dict={}
for i in range(1,16):
    my_dict[i]=i**2
print(my_dict)
```

```python
#6 WRITE A PROGRAM TO SUM ALL THE VALUES IN A DICTIONARY, CONSIDERING THE VALUES ALL IN THE INT TYPE
my_dict={"a":10,"b":20,"c":30,"d":40}
total=0
for value in my_dict.values():
    total=total+value
print("Sum of values:",total)
```

```python
#7 Assign 5 Elements in dictionary named my_dict. This dictionary contains student's information

# Create a dictionary with 5 students
my_dict = {
    1: {"name": "Drishti", "age": 18, "course": "BTECH"},
    2: {"name": "Sakshi", "age": 19, "course": "BTECH"},
    3: {"name": "Ishmeet", "age": 18, "course": "BTECH"},
    4: {"name": "Naman", "age": 19, "course": "BBA"},
    5: {"name": "Aditya", "age": 18, "course": "BTECH"}
}

# Add student information
my_dict[6] = {"name": "NeHAL", "age": 19, "course": "BSc"}

# Delete student information
del my_dict[3]

# Display student information
print("Student Information:")
for roll_no, student in my_dict.items():
    print("Roll No:", roll_no)
    print("Name:", student["name"])
    print("Age:", student["age"])
    print("Course:", student["course"])
    print()
```

```python
#10 FIND THE MEAN VALUES OF DICTIONARY
my_dict = {
    "A": 10,
    "B": 20,
    "C": 30,
    "D": 40,
    "E": 50
}

values = list(my_dict.values())

mean = sum(values) / len(values)

print("Mean =", mean)
```
