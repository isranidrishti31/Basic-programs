```python
#4 WRITE A PROGRAM TO ITERATE OVER A DICTIONARY USING FOR LOOP AND PRINT THE KEYS ALONE,ALONE AND BOTH KEY VALUES
my_dict={
    101:"Drishti",
    102:"Sakshi",
    103:"Ishmeet"
}
print("Keys:")
for key in my_dict:
    print (key)
for values in my_dict.values():
    print(values)
    print("\nKeys and values:")
for key,value in my_dict.items():
    print(key,value)
```

```python
#5 WRITE A PROGRAM TO PREPARE A DICTIONARY WHERE THE KEYS ARE NUMBERS BETWEEN 1 AND 15(BOTH INCLUDED) AND THE VALUES ARE SQUARES OF THE KEYS
my_dict={}
for i in range(1,16):
    my_dict[i]=i**2
print(my_dict)
```

```python
#6 WRITE A PROGRAM TO SUM ALL THE VALUES IN A DICTIONARY, CONSIDERING THE VALUES ALL IN THE INT TYPE
my_dict={"a":10,"b":20,"c":30,"d":40}
total=0
for value in my_dict.values():
    total=total+value
print("Sum of values:",total)
```

```python
#10 FIND THE MEAN VALUES OF DICTIONARY
my_dict = {
    "A": 10,
    "B": 20,
    "C": 30,
    "D": 40,
    "E": 50
}

values = list(my_dict.values())

mean = sum(values) / len(values)

print("Mean =", mean)
```

```python
#12 WRITE A PYTHON PROGRAM TO CREATE A DICTIONARY OF 5 STUDENTS WITH MARKS AND DISPLAY STUDENT WITH HIGHEST MARKS
students={
    "Drishti":100,
    "Sakshi":90,
    "Ishmeet":80,
    "Naman":70
}
highest_student=max(students,key=students.get)
print ("student with highest marks:",highest_student)
print("highest marks:",students[highest_student])
```

```python
#13 PROGRAM TO COUNT THE FREQUENCY OF EACH ELEMENT IN A LIST USING DICTIONARY
numbers=[10,20,30,10,20,10,50,40,40]
frequency={}
for num in numbers:
  if num in frequency:
   frequency[num]=frequency[num]+1
  else:
    frequency[num]=1
print("frequency of elements:",frequency)
```

```python
#14 PYTHON PROGRAM TO SORT A DICTIONARY BY ITS VALUES
my_dict={
     "Drishti":100,
    "Sakshi":90,
    "Ishmeet":80,
    "Naman":70
}
sorted_dict=dict(sorted(my_dict.items(),key=lambda item:item[1]))
print("dictionary sorted by values:")
print(sorted_dict)
```

```python
#14 PYTHON PROGRAM TO SORT A DICTIONARY BY ITS VALUES DESCENDING ORDER
my_dict={
     "Drishti":100,
    "Sakshi":90,
    "Ishmeet":80,
    "Naman":70
}
sorted_dict=dict(sorted(my_dict.items(),key=lambda item:item[1],
         reverse=True))
print(sorted_dict)
```
