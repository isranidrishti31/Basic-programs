```python
#9 Accept 20 values from the user and store them in a list

numbers = []

for i in range(20):
    value = int(input("Enter value " + str(i + 1) + ": "))
    numbers.append(value)

print("\nList entered by you:", numbers)

# a) Count similar elements and print their index values

print("\n--- Similar Elements and Their Index Values ---")

for value in set(numbers):
    indexes = []
    
    for i in range(len(numbers)):
        if numbers[i] == value:
            indexes.append(i)
    
    if len(indexes) > 1:
        print(value, "is repeated", len(indexes), "times at indexes:", indexes)


# b) Count even and odd values

even_count = 0
odd_count = 0

for value in numbers:
    if value % 2 == 0:
        even_count += 1
    else:
        odd_count += 1

print("\n--- Even and Odd Count ---")
print("Even values:", even_count)
print("Odd values:", odd_count)


# c) Count positive and negative values

positive_count = 0
negative_count = 0
zero_count = 0

for value in numbers:
    if value > 0:
        positive_count += 1
    elif value < 0:
        negative_count += 1
    else:
        zero_count += 1

print("\n--- Positive and Negative Count ---")
print("Positive values:", positive_count)
print("Negative values:", negative_count)
print("Zero values:", zero_count)
```

```python
#14 Create a list
my_list = [10, 20, 30, 40, 50]

print("Original list:", my_list)

# Insert an element
value = int(input("Enter value to insert: "))
position = int(input("Enter position: "))

my_list.insert(position, value)

print("List after insertion:", my_list)

# Delete an element
value = int(input("Enter value to delete: "))

if value in my_list:
    my_list.remove(value)
    print("List after deletion:", my_list)
else:
    print("Value not found in the list.")

# Display the final list
print("Final list:", my_list)
```

```python
#16 removing duplicates
my_list = [10, 20, 10, 30, 20, 40, 30, 50]

print("Original list:", my_list)

# Create an empty list
new_list = []

# Remove duplicates
for value in my_list:
    if value not in new_list:
        new_list.append(value)

print("List after removing duplicates:", new_list)
```
